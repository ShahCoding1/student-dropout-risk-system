```python
# ============================================================
#  STUDENT DROPOUT RISK ANALYTICS & EARLY WARNING SYSTEM
# ============================================================
# End-to-End Data Science Project
#
# Dataset:
# UCI - Predict Students' Dropout and Academic Success
#
# Main Tasks:
# 1. Data loading
# 2. Data quality analysis
# 3. Exploratory Data Analysis
# 4. Feature engineering
# 5. Statistical analysis
# 6. Full academic outcome prediction
# 7. Early-warning prediction
# 8. Model comparison
# 9. Cross-validation
# 10. Confusion matrices
# 11. Feature importance
# 12. Student risk scoring
# 13. At-risk student analysis
# 14. Export prediction results
# ============================================================


# ============================================================
# 1. INSTALL / IMPORT LIBRARIES
# ============================================================

!pip -q install ucimlrepo

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import chi2_contingency

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    LabelEncoder
)

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_auc_score
)

from sklearn.inspection import permutation_importance

from ucimlrepo import fetch_ucirepo

print("Libraries imported successfully! ")


# ============================================================
# 2. PROJECT TITLE
# ============================================================

print("=" * 80)
print(" STUDENT DROPOUT RISK ANALYTICS & EARLY WARNING SYSTEM")
print("=" * 80)


# ============================================================
# 3. LOAD OFFICIAL UCI DATASET
# ============================================================

print("\n Loading UCI dataset...")

dataset = fetch_ucirepo(id=697)

X = dataset.data.features.copy()
y = dataset.data.targets.copy()

df = pd.concat([X, y], axis=1)

print("Dataset loaded successfully! ")
print(f"Dataset shape: {df.shape}")


# ============================================================
# 4. BASIC DATA INSPECTION
# ============================================================

print("\n" + "=" * 80)
print(" DATASET OVERVIEW")
print("=" * 80)

print("\nFirst 5 rows:")
display(df.head())

print("\nRandom sample:")
display(df.sample(5, random_state=42))

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")

print("\nData types:")
display(df.dtypes)

print("\nDataset information:")
df.info()


# ============================================================
# 5. DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 80)
print(" DATA QUALITY CHECK")
print("=" * 80)

print("\nMissing values:")
missing = df.isnull().sum()
missing_df = pd.DataFrame({
    "Column": missing.index,
    "Missing Values": missing.values
})
display(missing_df[missing_df["Missing Values"] > 0])

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nUnique values per column:")
unique_df = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [df[col].nunique() for col in df.columns]
})
display(unique_df)


# ============================================================
# 6. IDENTIFY TARGET COLUMN
# ============================================================

target_candidates = [
    "Target",
    "target",
    "Outcome",
    "outcome"
]

target_col = None

for col in target_candidates:
    if col in df.columns:
        target_col = col
        break

if target_col is None:
    raise ValueError(
        "Target column not found. Available columns:\n"
        + str(df.columns.tolist())
    )

print("\n Target column:", target_col)

print("\nTarget distribution:")
target_counts = df[target_col].value_counts()

display(
    pd.DataFrame({
        "Count": target_counts,
        "Percentage": (target_counts / len(df) * 100).round(2)
    })
)


# ============================================================
# 7. TARGET DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

ax = sns.countplot(
    data=df,
    x=target_col,
    order=df[target_col].value_counts().index
)

plt.title("Student Academic Outcome Distribution")
plt.xlabel("Academic Outcome")
plt.ylabel("Number of Students")
plt.xticks(rotation=20)

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()


# ============================================================
# 8. TARGET PIE CHART
# ============================================================

plt.figure(figsize=(7, 7))

df[target_col].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Outcome Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()


# ============================================================
# 9. DUPLICATE CHECK / CLEANING
# ============================================================

df_clean = df.copy()

before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print(f"\nRows before duplicate removal: {before}")
print(f"Rows after duplicate removal:  {after}")
print(f"Duplicates removed: {before - after}")


# ============================================================
# 10. DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 80)
print(" DESCRIPTIVE STATISTICS")
print("=" * 80)

display(df_clean.describe().T)


# ============================================================
# 11. NUMERICAL FEATURES
# ============================================================

numeric_cols = df_clean.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

if target_col in numeric_cols:
    numeric_cols.remove(target_col)

print("\nNumber of numerical features:", len(numeric_cols))


# ============================================================
# 12. CATEGORICAL FEATURES
# ============================================================

categorical_cols = df_clean.select_dtypes(
    include=["object", "category"]
).columns.tolist()

if target_col in categorical_cols:
    categorical_cols.remove(target_col)

print("Number of categorical features:", len(categorical_cols))


# ============================================================
# 13. FEATURE ENGINEERING
# ============================================================

print("\n" + "=" * 80)
print("⚙️ FEATURE ENGINEERING")
print("=" * 80)

# The original dataset uses encoded numeric variables.
# We create interpretable derived indicators where possible.

df_model = df_clean.copy()

# ------------------------------------------------------------
# Academic performance features
# ------------------------------------------------------------

if "Curricular units 1st sem (approved)" in df_model.columns:
    df_model["Sem1_Approval_Rate"] = (
        df_model["Curricular units 1st sem (approved)"] /
        df_model["Curricular units 1st sem (enrolled)"].replace(0, np.nan)
    )

if "Curricular units 2nd sem (approved)" in df_model.columns:
    df_model["Sem2_Approval_Rate"] = (
        df_model["Curricular units 2nd sem (approved)"] /
        df_model["Curricular units 2nd sem (enrolled)"].replace(0, np.nan)
    )

if (
    "Curricular units 1st sem (grade)" in df_model.columns
    and "Curricular units 2nd sem (grade)" in df_model.columns
):
    df_model["Average_Semester_Grade"] = (
        df_model["Curricular units 1st sem (grade)"] +
        df_model["Curricular units 2nd sem (grade)"]
    ) / 2

# ------------------------------------------------------------
# Failed-unit indicator
# ------------------------------------------------------------

if (
    "Curricular units 1st sem (without evaluations)" in df_model.columns
    and "Curricular units 2nd sem (without evaluations)" in df_model.columns
):
    df_model["Total_Without_Evaluation"] = (
        df_model["Curricular units 1st sem (without evaluations)"] +
        df_model["Curricular units 2nd sem (without evaluations)"]
    )

# ------------------------------------------------------------
# Tuition indicator
# ------------------------------------------------------------

if "Tuition fees up to date" in df_model.columns:
    df_model["Tuition_At_Risk"] = (
        df_model["Tuition fees up to date"] == 0
    ).astype(int)

# ------------------------------------------------------------
# Scholarship indicator
# ------------------------------------------------------------

if "Scholarship holder" in df_model.columns:
    df_model["Scholarship_Status"] = (
        df_model["Scholarship holder"]
    ).map({
        0: "No Scholarship",
        1: "Scholarship"
    })

print("Feature engineering completed! ")


# ============================================================
# 14. DEMOGRAPHIC ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print(" DEMOGRAPHIC ANALYSIS")
print("=" * 80)

if "Age at enrollment" in df_model.columns:

    plt.figure(figsize=(10, 5))

    sns.histplot(
        df_model["Age at enrollment"],
        bins=20,
        kde=True
    )

    plt.title("Student Age Distribution at Enrollment")
    plt.xlabel("Age")
    plt.ylabel("Number of Students")

    plt.tight_layout()
    plt.show()


# ============================================================
# 15. AGE VS OUTCOME
# ============================================================

if "Age at enrollment" in df_model.columns:

    plt.figure(figsize=(10, 6))

    sns.boxplot(
        data=df_model,
        x=target_col,
        y="Age at enrollment"
    )

    plt.title("Age at Enrollment by Academic Outcome")
    plt.xlabel("Academic Outcome")
    plt.ylabel("Age at Enrollment")

    plt.tight_layout()
    plt.show()


# ============================================================
# 16. GENDER ANALYSIS
# ============================================================

if "Gender" in df_model.columns:

    gender_outcome = pd.crosstab(
        df_model["Gender"],
        df_model[target_col]
    )

    print("\nGender vs Outcome:")
    display(gender_outcome)

    gender_percentage = pd.crosstab(
        df_model["Gender"],
        df_model[target_col],
        normalize="index"
    ) * 100

    print("\nGender vs Outcome (%):")
    display(gender_percentage.round(2))

    gender_percentage.plot(
        kind="bar",
        figsize=(10, 6)
    )

    plt.title("Academic Outcome by Gender")
    plt.xlabel("Gender")
    plt.ylabel("Percentage")
    plt.xticks(rotation=0)
    plt.legend(title="Outcome")

    plt.tight_layout()
    plt.show()


# ============================================================
# 17. TUITION STATUS ANALYSIS
# ============================================================

if "Tuition fees up to date" in df_model.columns:

    tuition_outcome = pd.crosstab(
        df_model["Tuition fees up to date"],
        df_model[target_col],
        normalize="index"
    ) * 100

    print("\nTuition Status vs Outcome:")
    display(tuition_outcome.round(2))

    tuition_outcome.plot(
        kind="bar",
        figsize=(10, 6)
    )

    plt.title("Academic Outcome by Tuition Fee Status")
    plt.xlabel("Tuition Fees Up to Date")
    plt.ylabel("Percentage")
    plt.xticks(rotation=0)

    plt.tight_layout()
    plt.show()


# ============================================================
# 18. SCHOLARSHIP ANALYSIS
# ============================================================

if "Scholarship holder" in df_model.columns:

    scholarship_outcome = pd.crosstab(
        df_model["Scholarship holder"],
        df_model[target_col],
        normalize="index"
    ) * 100

    print("\nScholarship Status vs Outcome:")
    display(scholarship_outcome.round(2))

    scholarship_outcome.plot(
        kind="bar",
        figsize=(10, 6)
    )

    plt.title("Academic Outcome by Scholarship Status")
    plt.xlabel("Scholarship Holder")
    plt.ylabel("Percentage")
    plt.xticks(rotation=0)

    plt.tight_layout()
    plt.show()


# ============================================================
# 19. COURSE / PROGRAM ANALYSIS
# ============================================================

if "Course" in df_model.columns:

    course_outcome = pd.crosstab(
        df_model["Course"],
        df_model[target_col],
        normalize="index"
    ) * 100

    dropout_by_course = (
        course_outcome
        .sort_values(
            by="Dropout",
            ascending=False
        )
    )

    print("\nTop courses by dropout percentage:")
    display(dropout_by_course.head(10).round(2))

    plt.figure(figsize=(12, 7))

    dropout_by_course["Dropout"].head(10).sort_values().plot(
        kind="barh"
    )

    plt.title("Top 10 Courses by Dropout Percentage")
    plt.xlabel("Dropout Percentage")
    plt.ylabel("Course")

    plt.tight_layout()
    plt.show()


# ============================================================
# 20. ACADEMIC PERFORMANCE ANALYSIS
# ============================================================

academic_features = [
    "Curricular units 1st sem (approved)",
    "Curricular units 1st sem (grade)",
    "Curricular units 2nd sem (approved)",
    "Curricular units 2nd sem (grade)"
]

available_academic = [
    col for col in academic_features
    if col in df_model.columns
]

for col in available_academic:

    plt.figure(figsize=(10, 6))

    sns.boxplot(
        data=df_model,
        x=target_col,
        y=col
    )

    plt.title(f"{col} by Academic Outcome")
    plt.xlabel("Academic Outcome")
    plt.ylabel(col)

    plt.tight_layout()
    plt.show()


# ============================================================
# 21. SEMESTER GRADE ANALYSIS
# ============================================================

if "Curricular units 1st sem (grade)" in df_model.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        data=df_model,
        x="Curricular units 1st sem (grade)",
        hue=target_col,
        kde=True,
        element="step"
    )

    plt.title("First Semester Grade Distribution by Outcome")
    plt.xlabel("First Semester Grade")
    plt.ylabel("Number of Students")

    plt.tight_layout()
    plt.show()


if "Curricular units 2nd sem (grade)" in df_model.columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        data=df_model,
        x="Curricular units 2nd sem (grade)",
        hue=target_col,
        kde=True,
        element="step"
    )

    plt.title("Second Semester Grade Distribution by Outcome")
    plt.xlabel("Second Semester Grade")
    plt.ylabel("Number of Students")

    plt.tight_layout()
    plt.show()


# ============================================================
# 22. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print(" CORRELATION ANALYSIS")
print("=" * 80)

correlation_cols = [
    col for col in df_model.select_dtypes(include=np.number).columns
]

corr = df_model[correlation_cols].corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    linewidths=0.3
)

plt.title("Feature Correlation Matrix")

plt.tight_layout()
plt.show()


# ============================================================
# 23. CHI-SQUARE TEST FUNCTION
# ============================================================

def chi_square_test(data, feature, target):

    table = pd.crosstab(
        data[feature],
        data[target]
    )

    chi2, p, dof, expected = chi2_contingency(table)

    print("\n" + "-" * 70)
    print(f"Chi-Square Test: {feature} vs {target}")
    print("-" * 70)

    print(f"Chi-Square Statistic: {chi2:.4f}")
    print(f"Degrees of Freedom: {dof}")
    print(f"P-value: {p:.6f}")

    if p < 0.05:
        print("Result: Statistically significant association.")
    else:
        print("Result: No statistically significant association.")

    return p


# ============================================================
# 24. STATISTICAL TESTS
# ============================================================

categorical_test_features = [
    "Gender",
    "Tuition fees up to date",
    "Scholarship holder",
    "Daytime/evening attendance",
    "International"
]

for feature in categorical_test_features:

    if feature in df_model.columns:

        try:
            chi_square_test(
                df_model,
                feature,
                target_col
            )
        except Exception as e:
            print(f"Could not test {feature}: {e}")


# ============================================================
# 25. PREPARE DATA FOR MACHINE LEARNING
# ============================================================

print("\n" + "=" * 80)
print(" MACHINE LEARNING PREPARATION")
print("=" * 80)

ml_df = df_model.copy()

X_full = ml_df.drop(columns=[target_col])
y_full = ml_df[target_col].copy()

print("Full model feature count:", X_full.shape[1])
print("Target classes:")
print(y_full.value_counts())


# ============================================================
# 26. LABEL ENCODING TARGET
# ============================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y_full)

print("\nTarget encoding:")
for number, label in enumerate(label_encoder.classes_):
    print(f"{number} = {label}")


# ============================================================
# 27. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_full,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])


# ============================================================
# 28. PREPROCESSING PIPELINE
# ============================================================

numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ============================================================
# 29. DEFINE MODELS
# ============================================================

models = {

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        class_weight="balanced"
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8,
        min_samples_split=10,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=400,
        max_depth=12,
        min_samples_split=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=250,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    )
}


# ============================================================
# 30. TRAIN AND COMPARE MODELS
# ============================================================

print("\n" + "=" * 80)
print(" MODEL TRAINING & COMPARISON")
print("=" * 80)

results = []

trained_models = {}

for model_name, model in models.items():

    print(f"\nTraining {model_name}...")

    pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "model",
                model
            )
        ]
    )

    pipeline.fit(
        X_train,
        y_train
    )

    predictions = pipeline.predict(
        X_test
    )

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    try:

        probabilities = pipeline.predict_proba(
            X_test
        )

        roc_auc = roc_auc_score(
            y_test,
            probabilities,
            multi_class="ovr",
            average="weighted"
        )

    except Exception:

        roc_auc = np.nan

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": roc_auc
    })

    trained_models[model_name] = pipeline

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"ROC-AUC  : {roc_auc:.4f}")


# ============================================================
# 31. MODEL COMPARISON TABLE
# ============================================================

model_results = pd.DataFrame(results)

model_results = model_results.sort_values(
    by="F1 Score",
    ascending=False
)

print("\n" + "=" * 80)
print(" MODEL COMPARISON")
print("=" * 80)

display(
    model_results.style.format({
        "Accuracy": "{:.3f}",
        "Precision": "{:.3f}",
        "Recall": "{:.3f}",
        "F1 Score": "{:.3f}",
        "ROC-AUC": "{:.3f}"
    })
)


# ============================================================
# 32. MODEL COMPARISON VISUALIZATION
# ============================================================

plot_results = model_results.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1 Score"]
]

plot_results.plot(
    kind="bar",
    figsize=(13, 7)
)

plt.title("Machine Learning Model Comparison")
plt.ylabel("Score")
plt.xlabel("Model")
plt.ylim(0, 1)
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ============================================================
# 33. SELECT BEST MODEL
# ============================================================

best_model_name = model_results.iloc[0]["Model"]

best_model = trained_models[
    best_model_name
]

print("\n Best model based on weighted F1-score:")
print(best_model_name)


# ============================================================
# 34. BEST MODEL CLASSIFICATION REPORT
# ============================================================

best_predictions = best_model.predict(
    X_test
)

print("\n" + "=" * 80)
print(" BEST MODEL CLASSIFICATION REPORT")
print("=" * 80)

print(
    classification_report(
        y_test,
        best_predictions,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ============================================================
# 35. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    best_predictions
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)

plt.title(f"Confusion Matrix - {best_model_name}")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()
plt.show()


# ============================================================
# 36. CROSS-VALIDATION
# ============================================================

print("\n" + "=" * 80)
print(" CROSS-VALIDATION")
print("=" * 80)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = []

for model_name, pipeline in trained_models.items():

    scores = cross_val_score(
        pipeline,
        X_full,
        y_encoded,
        cv=cv,
        scoring="f1_weighted",
        n_jobs=-1
    )

    cv_results.append({
        "Model": model_name,
        "Mean CV F1": scores.mean(),
        "Std CV F1": scores.std()
    })

    print(
        f"{model_name}: "
        f"{scores.mean():.4f} ± {scores.std():.4f}"
    )

cv_df = pd.DataFrame(cv_results)

display(
    cv_df.sort_values(
        "Mean CV F1",
        ascending=False
    )
)


# ============================================================
# 37. FEATURE IMPORTANCE
# ============================================================

print("\n" + "=" * 80)
print(" FEATURE IMPORTANCE")
print("=" * 80)

# Tree-based model gives direct feature importance.
# If the selected model is not tree-based, use Random Forest.

importance_model_name = "Random Forest"

importance_model = trained_models[
    importance_model_name
]

rf_estimator = importance_model.named_steps["model"]
rf_preprocessor = importance_model.named_steps["preprocessor"]

feature_names = rf_preprocessor.get_feature_names_out()

importance_values = rf_estimator.feature_importances_

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance_values
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

print("\nTop 20 most important features:")

display(
    importance_df.head(20)
)


# ============================================================
# 38. FEATURE IMPORTANCE VISUALIZATION
# ============================================================

top_features = importance_df.head(20).sort_values(
    "Importance"
)

plt.figure(figsize=(12, 9))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.title(
    "Top 20 Features Influencing Student Outcome Prediction"
)

plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()


# ============================================================
# 39. EARLY WARNING MODEL
# ============================================================
#
# IMPORTANT:
# We now remove semester-performance variables.
#
# This creates a more realistic early-warning scenario using
# enrollment/background information rather than information
# observed after academic performance has already occurred.
# ============================================================

print("\n" + "=" * 80)
print(" EARLY WARNING MODEL")
print("=" * 80)

early_exclude = [
    # First semester performance
    "Curricular units 1st sem (credited)",
    "Curricular units 1st sem (enrolled)",
    "Curricular units 1st sem (evaluations)",
    "Curricular units 1st sem (approved)",
    "Curricular units 1st sem (grade)",
    "Curricular units 1st sem (without evaluations)",

    # Second semester performance
    "Curricular units 2nd sem (credited)",
    "Curricular units 2nd sem (enrolled)",
    "Curricular units 2nd sem (evaluations)",
    "Curricular units 2nd sem (approved)",
    "Curricular units 2nd sem (grade)",
    "Curricular units 2nd sem (without evaluations)",

    # Engineered semester features
    "Sem1_Approval_Rate",
    "Sem2_Approval_Rate",
    "Average_Semester_Grade",
    "Total_Without_Evaluation"
]

early_exclude = [
    col for col in early_exclude
    if col in ml_df.columns
]

X_early = ml_df.drop(
    columns=[target_col] + early_exclude
)

print("Removed late academic-performance features:")
for col in early_exclude:
    print(" -", col)

print("\nEarly-warning feature count:", X_early.shape[1])


# ============================================================
# 40. EARLY WARNING TRAIN / TEST SPLIT
# ============================================================

X_train_early, X_test_early, y_train_early, y_test_early = train_test_split(
    X_early,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)


# ============================================================
# 41. EARLY WARNING PREPROCESSOR
# ============================================================

early_numeric = X_train_early.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

early_categorical = X_train_early.select_dtypes(
    include=["object", "category"]
).columns.tolist()

early_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="median"
                    )
                ),
                (
                    "scaler",
                    StandardScaler()
                )
            ]),
            early_numeric
        ),

        (
            "cat",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="most_frequent"
                    )
                ),
                (
                    "onehot",
                    OneHotEncoder(
                        handle_unknown="ignore"
                    )
                )
            ]),
            early_categorical
        )
    ]
)


# ============================================================
# 42. EARLY WARNING RANDOM FOREST
# ============================================================

early_model = Pipeline(
    steps=[
        (
            "preprocessor",
            early_preprocessor
        ),
        (
            "model",
            RandomForestClassifier(
                n_estimators=500,
                max_depth=12,
                min_samples_split=5,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

early_model.fit(
    X_train_early,
    y_train_early
)

early_predictions = early_model.predict(
    X_test_early
)

early_probabilities = early_model.predict_proba(
    X_test_early
)


# ============================================================
# 43. EARLY WARNING MODEL PERFORMANCE
# ============================================================

early_accuracy = accuracy_score(
    y_test_early,
    early_predictions
)

early_precision = precision_score(
    y_test_early,
    early_predictions,
    average="weighted",
    zero_division=0
)

early_recall = recall_score(
    y_test_early,
    early_predictions,
    average="weighted",
    zero_division=0
)

early_f1 = f1_score(
    y_test_early,
    early_predictions,
    average="weighted",
    zero_division=0
)

early_auc = roc_auc_score(
    y_test_early,
    early_probabilities,
    multi_class="ovr",
    average="weighted"
)

print("\n" + "=" * 80)
print(" EARLY WARNING MODEL PERFORMANCE")
print("=" * 80)

print(f"Accuracy : {early_accuracy:.4f}")
print(f"Precision: {early_precision:.4f}")
print(f"Recall   : {early_recall:.4f}")
print(f"F1 Score : {early_f1:.4f}")
print(f"ROC-AUC  : {early_auc:.4f}")


# ============================================================
# 44. EARLY WARNING CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test_early,
        early_predictions,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ============================================================
# 45. EARLY WARNING CONFUSION MATRIX
# ============================================================

cm_early = confusion_matrix(
    y_test_early,
    early_predictions
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm_early,
    annot=True,
    fmt="d",
    cmap="Oranges",
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)

plt.title("Early Warning Model - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()
plt.show()


# ============================================================
# 46. EARLY WARNING FEATURE IMPORTANCE
# ============================================================

early_rf = early_model.named_steps["model"]

early_feature_names = (
    early_model
    .named_steps["preprocessor"]
    .get_feature_names_out()
)

early_importance = pd.DataFrame({
    "Feature": early_feature_names,
    "Importance": early_rf.feature_importances_
})

early_importance = early_importance.sort_values(
    "Importance",
    ascending=False
)

print("\nTop early-warning features:")

display(
    early_importance.head(20)
)


# ============================================================
# 47. EARLY WARNING FEATURE IMPORTANCE CHART
# ============================================================

top_early = early_importance.head(20).sort_values(
    "Importance"
)

plt.figure(figsize=(12, 9))

plt.barh(
    top_early["Feature"],
    top_early["Importance"]
)

plt.title(
    "Top Features in Early-Warning Student Risk Prediction"
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()


# ============================================================
# 48. STUDENT RISK SCORING
# ============================================================
#
# We calculate a dropout probability using the early-warning
# model.
# ============================================================

dropout_class_index = list(
    label_encoder.classes_
).index("Dropout")

all_early_probabilities = early_model.predict_proba(
    X_early
)

risk_df = ml_df[
    [
        col for col in [
            "Age at enrollment",
            "Gender",
            "Course",
            "Tuition fees up to date",
            "Scholarship holder",
            target_col
        ]
        if col in ml_df.columns
    ]
].copy()

risk_df["Dropout_Probability"] = (
    all_early_probabilities[:, dropout_class_index]
)

# Risk categories
def assign_risk(probability):

    if probability >= 0.70:
        return "High Risk"

    elif probability >= 0.40:
        return "Medium Risk"

    else:
        return "Low Risk"


risk_df["Risk_Level"] = (
    risk_df["Dropout_Probability"]
    .apply(assign_risk)
)

risk_df["Dropout_Probability_%"] = (
    risk_df["Dropout_Probability"] * 100
).round(2)

risk_df = risk_df.sort_values(
    "Dropout_Probability",
    ascending=False
)

print("\n" + "=" * 80)
print(" STUDENT RISK ANALYSIS")
print("=" * 80)

display(
    risk_df.head(20)
)


# ============================================================
# 49. RISK DISTRIBUTION
# ============================================================

risk_counts = risk_df["Risk_Level"].value_counts()

print("\nRisk distribution:")
display(
    pd.DataFrame({
        "Students": risk_counts,
        "Percentage": (
            risk_counts /
            len(risk_df) *
            100
        ).round(2)
    })
)


plt.figure(figsize=(9, 5))

ax = sns.countplot(
    data=risk_df,
    x="Risk_Level",
    order=[
        "High Risk",
        "Medium Risk",
        "Low Risk"
    ]
)

plt.title("Student Dropout Risk Distribution")
plt.xlabel("Risk Level")
plt.ylabel("Number of Students")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()


# ============================================================
# 50. HIGH-RISK STUDENTS
# ============================================================

high_risk_students = risk_df[
    risk_df["Risk_Level"] == "High Risk"
].copy()

print("\n🚨 HIGH-RISK STUDENTS")
print(
    f"Number of high-risk students: "
    f"{len(high_risk_students)}"
)

display(
    high_risk_students.head(25)
)


# ============================================================
# 51. RISK BY COURSE
# ============================================================

if "Course" in risk_df.columns:

    risk_by_course = pd.crosstab(
        risk_df["Course"],
        risk_df["Risk_Level"],
        normalize="index"
    ) * 100

    if "High Risk" in risk_by_course.columns:

        risk_by_course = risk_by_course.sort_values(
            "High Risk",
            ascending=False
        )

        print("\nTop courses by high-risk percentage:")

        display(
            risk_by_course.head(10).round(2)
        )


# ============================================================
# 52. RISK VS ACTUAL OUTCOME
# ============================================================

if target_col in risk_df.columns:

    risk_outcome = pd.crosstab(
        risk_df["Risk_Level"],
        risk_df[target_col],
        normalize="index"
    ) * 100

    print("\nRisk Level vs Actual Outcome:")
    display(
        risk_outcome.round(2)
    )

    risk_outcome.plot(
        kind="bar",
        figsize=(11, 6)
    )

    plt.title(
        "Actual Academic Outcome by Predicted Risk Level"
    )

    plt.xlabel("Predicted Risk Level")
    plt.ylabel("Percentage")
    plt.xticks(rotation=0)

    plt.tight_layout()
    plt.show()


# ============================================================
# 53. DROPOUT PROBABILITY DISTRIBUTION
# ============================================================

plt.figure(figsize=(11, 6))

sns.histplot(
    risk_df["Dropout_Probability"],
    bins=30,
    kde=True
)

plt.axvline(
    0.40,
    linestyle="--",
    label="Medium Risk Threshold"
)

plt.axvline(
    0.70,
    linestyle="--",
    label="High Risk Threshold"
)

plt.title(
    "Predicted Dropout Probability Distribution"
)

plt.xlabel("Predicted Dropout Probability")
plt.ylabel("Number of Students")

plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# 54. KPI SUMMARY
# ============================================================

total_students = len(df_clean)

dropout_count = (
    df_clean[target_col] == "Dropout"
).sum()

enrolled_count = (
    df_clean[target_col] == "Enrolled"
).sum()

graduate_count = (
    df_clean[target_col] == "Graduate"
).sum()

high_risk_count = (
    risk_df["Risk_Level"] == "High Risk"
).sum()

medium_risk_count = (
    risk_df["Risk_Level"] == "Medium Risk"
).sum()

low_risk_count = (
    risk_df["Risk_Level"] == "Low Risk"
).sum()

dropout_rate = (
    dropout_count /
    total_students *
    100
)

print("\n" + "=" * 80)
print(" EXECUTIVE STUDENT ANALYTICS DASHBOARD")
print("=" * 80)

print(f"""
Total Students          : {total_students:,}

Dropout Students        : {dropout_count:,}
Enrolled Students       : {enrolled_count:,}
Graduate Students       : {graduate_count:,}

Historical Dropout Rate : {dropout_rate:.2f}%

Predicted High Risk     : {high_risk_count:,}
Predicted Medium Risk   : {medium_risk_count:,}
Predicted Low Risk      : {low_risk_count:,}
""")


# ============================================================
# 55. AUTOMATED KEY FINDINGS
# ============================================================

print("\n" + "=" * 80)
print(" AUTOMATED KEY FINDINGS")
print("=" * 80)

print(
    f"1. The dataset contains {total_students:,} student records."
)

print(
    f"2. The historical dropout group contains "
    f"{dropout_count:,} students ({dropout_rate:.2f}%)."
)

if "Age at enrollment" in df_clean.columns:

    print(
        f"3. Average age at enrollment: "
        f"{df_clean['Age at enrollment'].mean():.2f} years."
    )

if "Curricular units 1st sem (grade)" in df_clean.columns:

    print(
        f"4. Average first-semester grade: "
        f"{df_clean['Curricular units 1st sem (grade)'].mean():.2f}."
    )

if "Curricular units 2nd sem (grade)" in df_clean.columns:

    print(
        f"5. Average second-semester grade: "
        f"{df_clean['Curricular units 2nd sem (grade)'].mean():.2f}."
    )

print(
    f"6. The early-warning model identifies "
    f"{high_risk_count:,} students as high risk under the "
    f"70% probability threshold."
)

print(
    f"7. The early-warning model weighted F1-score is "
    f"{early_f1:.3f}."
)

print(
    f"8. The early-warning model weighted ROC-AUC is "
    f"{early_auc:.3f}."
)

print(
    f"9. The strongest predictive features should be interpreted "
    f"as statistical predictors rather than direct causes of dropout."
)


# ============================================================
# 56. SAVE MODEL RESULTS
# ============================================================

model_results.to_csv(
    "student_dropout_model_comparison.csv",
    index=False
)

cv_df.to_csv(
    "student_dropout_cross_validation.csv",
    index=False
)

risk_df.to_csv(
    "student_dropout_risk_predictions.csv",
    index=False
)

importance_df.to_csv(
    "student_dropout_feature_importance.csv",
    index=False
)

early_importance.to_csv(
    "student_early_warning_feature_importance.csv",
    index=False
)

print("\n Exported model analysis files:")
print("✓ student_dropout_model_comparison.csv")
print("✓ student_dropout_cross_validation.csv")
print("✓ student_dropout_risk_predictions.csv")
print("✓ student_dropout_feature_importance.csv")
print("✓ student_early_warning_feature_importance.csv")


# ============================================================
# 57. DOWNLOAD MAIN RESULTS
# ============================================================

try:

    from google.colab import files

    files.download(
        "student_dropout_risk_predictions.csv"
    )

except Exception:

    print(
        "\nGoogle Colab download is unavailable "
        "outside Colab."
    )


# ============================================================
# 58. FINAL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 80)
print(" PROJECT COMPLETED SUCCESSFULLY")
print("=" * 80)

print("""
Student Dropout Risk Analytics & Early Warning System

✓ Dataset loaded
✓ Data quality checked
✓ Exploratory analysis completed
✓ Feature engineering completed
✓ Statistical analysis completed
✓ Multiple ML models trained
✓ Model comparison completed
✓ Cross-validation completed
✓ Confusion matrix generated
✓ Feature importance analyzed
✓ Early-warning model developed
✓ Student dropout probabilities calculated
✓ Risk levels assigned
✓ High-risk students identified
✓ Results exported

""")


# ============================================================
# 59. OPTIONAL: CREATE A FINAL PORTFOLIO SUMMARY TABLE
# ============================================================

portfolio_summary = pd.DataFrame({
    "Metric": [
        "Total Students",
        "Dropout Students",
        "Graduate Students",
        "Enrolled Students",
        "Historical Dropout Rate",
        "High Risk Students",
        "Medium Risk Students",
        "Low Risk Students",
        "Best Full Model",
        "Early Warning F1",
        "Early Warning ROC-AUC"
    ],

    "Value": [
        total_students,
        dropout_count,
        graduate_count,
        enrolled_count,
        f"{dropout_rate:.2f}%",
        high_risk_count,
        medium_risk_count,
        low_risk_count,
        best_model_name,
        f"{early_f1:.3f}",
        f"{early_auc:.3f}"
    ]
})

print("\n FINAL PORTFOLIO SUMMARY")

display(portfolio_summary)



# ============================================================
#  WARNING MODEL FOR WEB APPLICATION
# ============================================================

import joblib
import json
import os

# Save trained early-warning pipeline
joblib.dump(
    early_model,
    "student_dropout_early_warning_model.joblib"
)

# Save target classes
model_metadata = {
    "target_classes": label_encoder.classes_.tolist(),
    "dropout_class": "Dropout",
    "risk_thresholds": {
        "low": 0.40,
        "medium": 0.70,
        "high": 1.00
    }
}

with open(
    "model_metadata.json",
    "w"
) as f:
    json.dump(
        model_metadata,
        f,
        indent=4
    )

print(" Model saved:")
print("student_dropout_early_warning_model.joblib")

print("\n Metadata saved:")
print("model_metadata.json")
```
